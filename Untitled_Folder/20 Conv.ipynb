Сделаем свой генератор для датасета

Про `PyDataset`/`Sequence` больше [тут](https://colab.research.google.com/corgiredirector?site=https%3A%2F%2Fwww.tensorflow.org%2Fapi_docs%2Fpython%2Ftf%2Fkeras%2Futils%2FSequence)

````python
from tensorflow.keras.utils import Sequence


class DataGenerator(Sequence):
    def __init__(self, paths, labels, batch_size=32):
        self.batch_size = batch_size
        self.paths = paths
        self.labels = labels

    def __len__(self):
        return int(np.round(len(self.paths) / self.batch_size))

    def __getitem__(self, index):
        X = self.paths[index * self.batch_size : (index+1) * self.batch_size]
        y = self.labels[index * self.batch_size : (index+1) * self.batch_size]
        X = np.array([self.load_img(i) for i in X])
        return X, np.array(y)

    @staticmethod
    def load_img(path):
        img = plt.imread(path)
        img = img / 255.0
        return img
````

````python
train_datagen = DataGenerator(train_paths, train_classes)
test_datagen = DataGenerator(test_paths, test_classes)
````

## Свертка

Обсудить эту идею можно с помощью нашего мозга. Нейроны или определенные регионы человеческого мозга [активируются](https://www.researchgate.net/publication/8491620_The_human_visual_cortex) на линии, формы, объекты, лица.

Значит нужно получить такой инструмент, который сможет имитировать поведение человеческого мозга и активироваться на определенные признаки (паттерны) на изображении.

````python
def apply_filter_example(img, kernel):
    size_x = img.shape[1]
    size_y = img.shape[0]
    img_transformed = np.zeros((3, 3))
    for x in range(1, size_x-1):
        for y in range(1, size_y-1):
            convolution = 0.0

            dimg = img[x-1:x+2, y-1:y+2]
            convolution = np.sum(dimg * kernel)

            img_transformed[x-1, y-1] = convolution
    return img_transformed


img_transformed = apply_filter_example(img, kernel)

fig, axs = plt.subplots(1, 2, figsize=(10, 5))
sns.heatmap(img, cmap='gray', annot=True, cbar=False, ax=axs[0])
sns.heatmap(img_transformed, cmap='gray', annot=True, cbar=False, linewidths=1, linecolor='k', ax=axs[1]);
````

**Свертка**

````python
def apply_filter(img, kernel):
    size_x = img.shape[0]
    size_y = img.shape[1]
    img_transformed = np.copy(img)
    for x in range(1, size_x-1):
        for y in range(1, size_y-1):
            convolution = 0.0

            dimg = img[x-1:x+2, y-1:y+2]
            convolution = np.sum(dimg * kernel)

            if convolution < 0:
                convolution = 0
            if convolution > 255:
                convolution = 255
            img_transformed[x, y] = convolution
    return img_transformed
````

````python
def plot_image(img, img_transformed):

    fig, axs = plt.subplots(2, 2, figsize=(15, 7))
    axs[0][0].imshow(img, cmap='gray')
    axs[0][1].imshow(img_transformed, cmap='gray')

    ymin,ymax,xmin,xmax = 25,75,150,200

    axs[1][0].imshow(img[ymin:ymax,xmin:xmax], cmap='gray')
    axs[1][1].imshow(img_transformed[ymin:ymax,xmin:xmax], cmap='gray')
    plt.show()
````

Параметр `padding` отвечает за то, каким образом мы заполняем границы изображения при применении свертки. Существуют два типа:
- `padding='valid'` (без заполнения)
- `padding='same'` (заполнение таким образом, чтобы размер выходного изображения был таким же, как размер входного).

#### stride

Параметр `stride` определяет расстояние между применениями ядра к изображению. Увеличение значения stride уменьшает размер выходного изображения, что помогает ускорить вычисления и уменьшить размер данных.

Более большой `stride` может привести к более абстрактным признакам, а меньший `stride` может выделять более детальные признаки.

#### MaxPooling

````python
from keras import Sequential
from keras.layers import Conv2D, MaxPooling2D

fig, ax = plt.subplots(3, 2, figsize=(12, 14))

for i, s in enumerate(np.arange(2, 5)):
    random.seed(1)
    tf.random.set_seed(1)
    model_conv = Sequential([
        Conv2D(filters=1,
            kernel_size=(3, 3),
            use_bias=False,
            padding='same',
            strides=(s, s),
            input_shape=(img_bw.shape[0], img_bw.shape[1], 1))
    ])
    amount_w = model_conv.layers[0].get_weights()[0].size

    pred = model_conv.predict(img_bw[None], verbose=0)[0]
    ax[i][0].set_title(f'weights {amount_w}\nstrides ({s}, {s}). img size {pred.shape}')
    ax[i][0].imshow(pred, cmap='gray')

    model_pool = Sequential([
        MaxPooling2D(pool_size=(2, 2), strides=(s, s),
                     input_shape=(img_bw.shape[0], img_bw.shape[1], 1))
    ])
    # amount_w = model_pool.layers[0].get_weights()

    pred = model_pool.predict(img_bw[None], verbose=0)[0]
    ax[i][1].set_title(f'weights 0\nstrides ({s}, {s}). img size {pred.shape}')
    ax[i][1].imshow(pred, cmap='gray')
````

````python
random.seed(9)
tf.random.set_seed(9)

input = Input(shape=(256, 256, 3))

x = Conv2D(filters=3, kernel_size=(3, 3), activation='relu',
           padding='same', strides=(1, 1), dilation_rate=(1, 1))(input)
x = MaxPooling2D()(x)
x = Conv2D(filters=6, kernel_size=(3, 3), activation='relu',
           padding='same', strides=(1, 1), dilation_rate=(1, 1))(x)
x = MaxPooling2D()(x)
x = Flatten()(x)
output = Dense(3, activation='softmax')(x)

model_max_pool = Model(inputs=input, outputs=output)
model_max_pool.summary()
````

А теперь размерность изображения потихоньку уменьшается, из-за этого будет меньше сигналов на выходе из сверточных слоев, а значит будет меньше параметров:
