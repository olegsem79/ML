```python
from ultralytics import YOLO
import glob
import os
import pandas as pd
from tqdm import tqdm

# Путь к папке с моделями
models_folder = "/home/oleg/projects/ML/COMPARE/Pretrain_model_UAV"
data_yaml = "/home/oleg/projects/ML/COMPARE/UAV-6_YOLO11/data.yaml"

# Находим все .pt файлы
model_files = glob.glob(os.path.join(models_folder, "*.pt"))
print(f"🔍 Найдено моделей: {len(model_files)}")

# Результаты
results_data = []

print("🚀 Начинаем сравнивать модели на тестовой выборке...")

for model_path in tqdm(model_files, desc="Тестирование моделей"):
    try:
        model_name = os.path.basename(model_path)
        print(f"\n🧪 Тестируем: {model_name}")
        
        # Загружаем модель
        model = YOLO(model_path)
        
        # Валидация на тестовой выборке
        results = model.val(
            data=data_yaml,
            split='test',
            imgsz=640,
            batch=32,
            device=0,
            verbose=False  # Отключаем подробный вывод для каждой модели
        )
        
        # Сохраняем результаты
        model_results = {
            'Model': model_name,
            'mAP50': results.box.map50,
            'mAP50-95': results.box.map,
            'Precision': results.box.mp,
            'Recall': results.box.mr,
            'Params': sum(p.numel() for p in model.model.parameters()),
            'Model_Size_MB': os.path.getsize(model_path) / 1024 / 1024
        }
        
        results_data.append(model_results)
        
        print(f"✅ {model_name}: mAP50={results.box.map50:.4f}")
        
    except Exception as e:
        print(f"❌ Ошибка с {model_path}: {e}")
        continue

# Создаем DataFrame с результатами
if results_data:
    df = pd.DataFrame(results_data)
    
    # Сортируем по mAP50 (лучшие модели сначала)
    df = df.sort_values('mAP50', ascending=False)
    
    print("\n" + "="*80)
    print("🎯 РЕЗУЛЬТАТЫ СРАВНЕНИЯ МОДЕЛЕЙ")
    print("="*80)
    
    # Красиво форматируем вывод
    pd.set_option('display.float_format', '{:.4f}'.format)
    print(df.to_string(index=False))
    
    # Сохраняем в CSV
    output_csv = "/home/oleg/projects/ML/COMPARE/model_comparison_results.csv"
    df.to_csv(output_csv, index=False)
    print(f"\n💾 Результаты сохранены в: {output_csv}")
    
else:
    print("❌ Не удалось протестировать ни одну модель")
```

```python
from ultralytics import YOLO
import glob
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

models_folder = "/home/oleg/projects/ML/COMPARE/Pretrain_model_UAV"
data_yaml = "/home/oleg/projects/ML/COMPARE/UAV-6_YOLO11/data.yaml"

model_files = glob.glob(os.path.join(models_folder, "*.pt"))
print(f"🔍 Найдено моделей: {len(model_files)}")

results_data = []

for model_path in model_files:
    try:
        model_name = os.path.basename(model_path)
        print(f"🧪 {model_name}")
        
        model = YOLO(model_path)
        results = model.val(data=data_yaml, split='test', verbose=False)
        
        results_data.append({
            'Model': model_name.replace('.pt', ''),
            'mAP50': results.box.map50,
            'mAP50-95': results.box.map,
            'Precision': results.box.mp,
            'Recall': results.box.mr
        })
        
    except Exception as e:
        print(f"❌ {model_name}: {e}")
        continue

# Анализ результатов
if results_data:
    df = pd.DataFrame(results_data)
    df = df.sort_values('mAP50', ascending=False)
    
    # Визуализация
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # mAP50 сравнение
    sns.barplot(data=df, x='mAP50', y='Model', ax=axes[0,0], palette='viridis')
    axes[0,0].set_title('mAP50 Comparison')
    axes[0,0].set_xlim(0, 1)
    
    # mAP50-95 сравнение
    sns.barplot(data=df, x='mAP50-95', y='Model', ax=axes[0,1], palette='plasma')
    axes[0,1].set_title('mAP50-95 Comparison')
    axes[0,1].set_xlim(0, 1)
    
    # Precision-Recall scatter
    axes[1,0].scatter(df['Precision'], df['Recall'], s=100, alpha=0.7)
    for i, row in df.iterrows():
        axes[1,0].annotate(row['Model'], (row['Precision'], row['Recall']), 
                          xytext=(5, 5), textcoords='offset points', fontsize=8)
    axes[1,0].set_xlabel('Precision')
    axes[1,0].set_ylabel('Recall')
    axes[1,0].set_title('Precision vs Recall')
    axes[1,0].grid(True, alpha=0.3)
    
    # Heatmap метрик
    metrics_df = df.set_index('Model')[['mAP50', 'mAP50-95', 'Precision', 'Recall']]
    sns.heatmap(metrics_df, annot=True, cmap='YlOrRd', ax=axes[1,1])
    axes[1,1].set_title('Metrics Heatmap')
    
    plt.tight_layout()
    plt.show()
    
    print("\n📊 ТОП-3 модели по mAP50:")
    print(df.head(3).to_string(index=False))
```
